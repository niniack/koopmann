```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
rs = np.random.RandomState(365)
values = rs.randn(365, 4).cumsum(axis=0)
dates = pd.date_range("1 1 2016", periods=365, freq="D")
data = pd.DataFrame(values, dates, columns=["A", "B", "C", "D"])
data = data.rolling(7).mean()
```

```python
axes = sns.lineplot(data=data)
```

```python
import koopmann.aesthetics  # NOQA: F401
from koopmann.aesthetics import kill_axes, kill_ticks, palette, set_spine_color
```

```python
axes = sns.lineplot(data=data, palette=palette)
```

```python
axes = sns.lineplot(data=data, palette=palette)
set_spine_color(axes, "lightgray")
kill_ticks(axes)
kill_axes(axes)
```
